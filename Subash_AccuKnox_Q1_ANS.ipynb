Q1 : API Data Retrieval and Storage: You are tasked with fetching data from an external REST API, storing it in a local SQLite database, and displaying the retrieved data. The API provides a list of books in JSON format with attributes like title, author, and publication year.


```python

#Step 1 : Necessary imports and fetching from open library API
import requests
import sqlite3
URL = "https://openlibrary.org/search.json?q=python&limit=10&fields=title,author_name,first_publish_year"
books = requests.get(URL).json()["docs"]

```

```python
# Step 2: create database and table
conn = sqlite3.connect("books.db")
cur = conn.cursor()
cur.execute("CREATE TABLE IF NOT EXISTS books (id INTEGER PRIMARY KEY AUTOINCREMENT, title TEXT, author TEXT, year INTEGER)")
cur.execute("DELETE FROM books")


```

```python
# Step 3: store the books
for b in books:
    title = b.get("title", "Unknown")
    author = ", ".join(b.get("author_name", ["Unknown"]))
    year = b.get("first_publish_year")
    cur.execute("INSERT INTO books (title, author, year) VALUES (?, ?, ?)", (title, author, year))

conn.commit()

```

```python
# Step 4: display the data
print("\nBooks in database:")
for row in cur.execute("SELECT * FROM books"):
    print(row)

conn.close()
```
