```python
import requests
import random
import matplotlib.pyplot as plt

```

```python
# Step 1: fetch the data
URL = "https://jsonplaceholder.typicode.com/users"
users = requests.get(URL, timeout=10).json()

random.seed(1)  # keeps the scores the same every run
students = []
for u in users:
    students.append({"name": u["username"], "score": random.randint(40, 100)})

print(students)

```

```python
# Step 2: calculate the average
total = 0
for s in students:
    total = total + s["score"]
average = total / len(students)
print("Average score:", round(average, 2))

```

```python
# Step 3: bar chart
names = [s["name"] for s in students]
scores = [s["score"] for s in students]

plt.figure(figsize=(10, 5))
plt.bar(names, scores, color="skyblue")
plt.axhline(average, color="red", linestyle="--", label=f"Average: {average:.1f}")
plt.title("Student Test Scores")
plt.xlabel("Student")
plt.ylabel("Score")
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```
