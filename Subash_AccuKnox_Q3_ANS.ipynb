```python
import csv
import sqlite3

```

```python
# open the csv file & print users
with open("users.csv", newline="") as f:
    users = list(csv.DictReader(f))

print(users)

```

```python
# create users.db using sqlite3 and creating a table
conn = sqlite3.connect("users.db")
cur = conn.cursor()
cur.execute("CREATE TABLE IF NOT EXISTS users (id INTEGER PRIMARY KEY AUTOINCREMENT, name TEXT, email TEXT)")
cur.execute("DELETE FROM users")  # start fresh each run, no duplicates


```

```python
#inserting users into the db table
for u in users:
    cur.execute("INSERT INTO users (name, email) VALUES (?, ?)", (u["name"], u["email"]))

conn.commit()
print("Inserted", len(users), "users into users.db")

```

```python
# displaying the users
for row in cur.execute("SELECT * FROM users"):
    print(row)

conn.close()
```
